```python
epsilon      = 0.5
p_2          = 2.9919245059286
k            = 1.4e-3
omega_square = 0.25 * (1 - epsilon) + k / 6
p_1          = k / (6.0 * omega_square * p_2)
r            = lambda t : 1 + 0.5 * (1 + t) * (p_2 - 1)
g            = lambda x : 2 * omega_square * (1 - p_1 / x)
f            = lambda t : 1 / np.sqrt(2 * g(r(t)))
p            = lambda t: f(t) / np.sqrt(1.0 - t**2)

a = -1
b = 1
reference_solution,_ = quad(p, a, b)
```
