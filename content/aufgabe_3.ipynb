Als tatsächlichen numerischen Test vergleichen wir nun alle drei Verfahren mit verschiedenen Testfunktionen. Sie können gerne auch mit anderen Funktionen $f$ experimentieren.

```python
from util.plotting_03 import setup_plotting, add_to_plot

f1 = lambda t, y : np.sign(t)
analytical_solution1 = lambda t, t0, y0 : y0 + abs(t) - abs(t0)

t0, t1 = -1.0, 1.0
y0 = 1.0

fig = setup_plotting("y'(t) = sign(t)", figsize=(15,8))
symbols = ["*", "o", ".", "x", "+"] # different markers for different discretizations

for i, n_steps in enumerate([10, 20, 40, 80]):
    # Setup
    h = (t1 - t0) / n_steps
    ts = np.linspace(t0, t1, n_steps + 1)
    ys_Euler, ys_Euler_modifiziert, ys_Heun = np.zeros(n_steps + 1), np.zeros(n_steps + 1), np.zeros(n_steps + 1)

    # Computation
    ys_Euler[0], ys_Euler_modifiziert[0], ys_Heun[0] = y0, y0, y0
    for j, t in enumerate(ts[:-1]):
        ys_Euler[j+1] = Euler(f1, t, ys_Euler[j], h)
        ys_Euler_modifiziert[j+1] = Euler_modifiziert(f1, t, ys_Euler_modifiziert[j], h)
        ys_Heun[j+1]  = Heun(f1, t, ys_Heun[j], h)

    # Reference Solution
    ys_true = np.zeros(n_steps + 1)
    for j, t in enumerate(ts):
        ys_true[j] = analytical_solution1(t, t0, y0)

    # Error Calculation
    errors_Euler = np.abs(ys_true - ys_Euler)
    errors_Euler_modifiziert = np.abs(ys_true - ys_Euler_modifiziert)
    errors_Heun = np.abs(ys_true - ys_Heun)

    # Plotting
    symbol = symbols[i]
    add_to_plot(fig, ts, ys_Euler, errors_Euler, ys_Euler_modifiziert, errors_Euler_modifiziert, ys_Heun, errors_Heun, symbol=symbol)
    
```
